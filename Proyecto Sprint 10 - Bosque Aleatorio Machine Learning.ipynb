## Preparación y Division de los datos


```python
# Separar las características y el objetivo

features = df.drop(['is_ultra'], axis=1)
target = df['is_ultra']

# Revisar las dimensiones
print("Dimensiones de features:", features.shape)
print("Dimensiones de target:", target.shape)
```

```python


from sklearn.model_selection import train_test_split

#Primera división: 60% entrenamiento y 40% temporal 
features_train, features_temp,target_train, target_temp = train_test_split(
    features,
    target,
    test_size=0.40,
    random_state=12345,
    stratify=target
)

#Segunda division :20% validacion y el 20 % prueba
features_valid, features_test, target_valid, target_test = train_test_split(
    features_temp,
    target_temp,
    test_size = 0.50,
    random_state=12345,
    stratify=target_temp
)

# Comprobar los tamaños de cada conjunto
print("Datos originales:", len(df))
print("Entrenamiento:", len(features_train))
print("Validación:", len(features_valid))
print("Prueba:", len(features_test))
```

```python
# Verifaicar la distribución de los tres conjuntos
print("Distribución del conjunto de entrenamiento:")
print(target_train.value_counts(normalize=True))

print("\nDistribución del conjunto de validación:")
print(target_valid.value_counts(normalize=True))

print("\nDistribución del conjunto de prueba:")
print(target_test.value_counts(normalize=True))
```

# Entrenamiento y comparación de modelos


```python

# Importar los modelos de clasificación
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# Importar la métrica de evaluación
from sklearn.metrics import accuracy_score
```

## Árbol de Decisión
Este algoritmo aprende reglas a partir de las caracteristicas de los usuarios para determinar que recomendar.

```python
# Entrenar un árbol de decisión con diferentes profundidades

best_accuracy_tree = 0
best_depth_tree = 0

for depth in range(1, 11):

    model_tree = DecisionTreeClassifier(
        max_depth=depth,
        random_state=12345
    )

    # Entrenar el modelo
    model_tree.fit(features_train, target_train)

    # Realizar predicciones sobre el conjunto de validación
    predictions_valid = model_tree.predict(features_valid)

    # Calcular la exactitud
    accuracy = accuracy_score(target_valid, predictions_valid)

    print("Profundidad:", depth, "| Exactitud:", round(accuracy, 4))

    # Guardar el mejor resultado
    if accuracy > best_accuracy_tree:
        best_accuracy_tree = accuracy
        best_depth_tree = depth

print("\nMejor resultado del árbol de decisión:")
print("Mejor profundidad:", best_depth_tree)
print("Mejor exactitud:", round(best_accuracy_tree, 4))
```

## Segundo Modelo: Bosque aleatorio

Este algoritmo construye varios arboles de decision y combina sus resultados para producir una clasificación

```python
# Entrenar un bosque aleatorio con diferentes hiperparámetros

best_accuracy_forest = 0
best_est_forest = 0
best_depth_forest = 0

for est in range(10, 101, 10):

    for depth in range(1, 11):

        model_forest = RandomForestClassifier(
            random_state=12345,
            n_estimators=est,
            max_depth=depth
        )

        # Entrenar el modelo
        model_forest.fit(features_train, target_train)

        # Realizar predicciones
        predictions_valid = model_forest.predict(features_valid)

        # Calcular exactitud
        accuracy = accuracy_score(target_valid, predictions_valid)

        print(
            "Árboles:", est,
            "| Profundidad:", depth,
            "| Exactitud:", round(accuracy, 4)
        )

        # Guardar la mejor configuración
        if accuracy > best_accuracy_forest:
            best_accuracy_forest = accuracy
            best_est_forest = est
            best_depth_forest = depth

print("\nMejor resultado del bosque aleatorio:")
print("Cantidad de árboles:", best_est_forest)
print("Profundidad:", best_depth_forest)
print("Mejor exactitud:", round(best_accuracy_forest, 4))
```

## Tercer Modelo : Regresión logistica 

A diferencia de los dos modelos anteriores, este algoritmo busca una relación matemática entre las caracteristicas y la probabilidad de pertenecer a una categoria

```python
# Entrenar un modelo de regresión logística 

model_logistic = LogisticRegression(
    random_state=12345,
    max_iter=1000
)

# Entrenar el modelo 
model_logistic.fit(features_train, target_train)

# Realizar predicciones
predictions_logistic = model_logistic.predict(features_valid)

# Calcular exactitud 
accuracy_logistic = accuracy_score(
    target_valid,
    predictions_logistic
)

print("Exactitud de la regresion logística:", round(accuracy_logistic))
```

## Comparar los tres modelos
Ahora que tenemos los resultados de los tres algoritmos, vamos a construir una tabla comparativa.

```python
resultados = pd.DataFrame({
    'Modelo': [
        'Árbol de decisión',
        'Bosque aleatorio',
        'Regresión logística'
    ],
    'Exactitud': [
        best_accuracy_tree,
        best_accuracy_forest,
        accuracy_logistic
    ]
})

# Ordenar los modelos de mayor a menor exactitud
resultados = resultados.sort_values(
    by='Exactitud',
    ascending= False
)

display(resultados)
```

### Conclusión del Estudio de Modelos

Durante el entrenamiento y evaluación de los modelos de clasificación se probaron tres algoritmos: árbol de decisión, bosque aleatorio y regresión logística. El bosque aleatorio obtuvo el mejor resultado, con una exactitud de 0.821151(82.12%), seguido del árbol de decisión con 0.802448(80.25%). La regresión logística obtuvo una exactitud de 0.738725(73.87%), por debajo del umbral mínimo establecido por Megaline. Por esta razón, se seleccionó el bosque aleatorio como el modelo candidato para la evaluación final.

# Evaluación final con el conjunto de prueba

Ahora vamos a comprobar si el modelo mantiene su buen rendimiento con datos que no utilizamos para entrenarlo ni para seleccionar sus hiperparametros.


### Preparar los datos para el entrenamiento final

```python
# Unir los conjuntos de entrenamiento y validación

features_train_valid = pd.concat(
    [features_train, features_valid]
)

target_train_valid = pd.concat(
    [target_train, target_valid]
)

# Comprobar las dimensiones
print("Características para entrenamiento final:", features_train_valid.shape)
print("Objetivo para entrenamiento final:", target_train_valid.shape)

print("Características para prueba:", features_test.shape)
print("Objetivo para prueba:", target_test.shape)
```

 #### Entrenar el mejor modelo
 Vamos a utilizar los mejores hiperparámetros que encontraste en el bosque aleatorio durante el paso anterior

```python
# Crear el modelo final con los mejores hiperparametros
final_model = RandomForestClassifier(
    random_state=12345,
    n_estimators=best_est_forest,
    max_depth=best_depth_forest
)

# Entrenar con el 80% de los datos
final_model.fit(features_train_valid, target_train_valid)

# Realizar predicciones sobre el conjunto de prueba
predictions_test = final_model.predict(features_test)

#Calcular la exactitud final
accuracy_test = accuracy_score(target_test, predictions_test)

print("Exactitud final del modelo:", round(accuracy_test,4))
```

### Conclusión para tu proyecto 
Después de seleccionar el modelo de bosque aleatorio, se realizó el entrenamiento final utilizando los conjuntos de entrenamiento y validación. Posteriormente, se evaluó su rendimiento con el conjunto de prueba, obteniendo una exactitud de 0.8289(82.89%).Este resultado supera el umbral mínimo de 0.75 establecido por Megaline. Por lo tanto, el modelo cumple con el objetivo de clasificación y demuestra una buena capacidad para recomendar los planes Smart y Ultra.

 ## Prueba de Cordura (Sanity Check)
La finalidad es comprobar que nuestro modelo es realemente util y que no esta obteniendo buenos resultados simplemente porque predice siempre el plan más frecuente

### Crear un modelo de referencia
Vamos a utilizar DummyClassifier, un clasificador sencillo que no aprende patrones de comportamiento. En este caso, siempre predecirá la categoría más frecuente.

```python
# Importar el modelo de referencia
from sklearn.dummy import DummyClassifier

# Crear el modelo que siempre predice la clase más frecuente
dummy_model = DummyClassifier(strategy='most_frequent')

# Entrenar el modelo de referencia
dummy_model.fit(features_train_valid, target_train_valid)

# Realizar predicciones sobre el conjunto de prueba
dummy_predictions = dummy_model.predict(features_test)

# Calcular la exactitud del modelo de referencia 
dummy_accuracy = accuracy_score(target_test, dummy_predictions)

print("Exactitud del modelo de referencia:", round(dummy_accuracy, 4))
print("Exactitud del bosque aleatorio:", round(accuracy_test, 4))
```

 ## Comparar los resultados
Ahora vamos a comprobar si nuestro modelo supera al modelo de referencia.

```python
if accuracy_test > dummy_accuracy:
    print("Prueba de cordura superada.")
    print("El modelo de bosque aleatorio es mejor que la predicción más frecuente.")
else:
    print("Prueba de cordura no superada.")
    print("El modelo no supera al modelo de referencia.")
```
